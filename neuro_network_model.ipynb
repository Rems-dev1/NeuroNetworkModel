```python
import torch
import torch.nn as nn
import torch.nn.functional as F # this helps to move data forward in layers
```

```python
# create a model class that inherits nn.module

class Model(nn.Module):
  # input layer (4 features of a flower) -> hidden layer 1(number of neurons) --> H2 (n) --> ouput (3 clases of flower)
  def __init__(self, in_feature = 4, h1 = 8,h2=9, out_feature = 3):
    super().__init__()
    # create layers
    self.fc1 = nn.Linear(in_feature, h1)
    self.fc2 = nn.Linear(h1, h2)
    self.out = nn.Linear(h2, out_feature)

  def forward(self, x):
    x = F.relu(self.fc1(x))
    x = F.relu(self.fc2(x))
    x = self.out(x)
    return x
```

```python
# this will tell it to start from 41 for randomization
torch.manual_seed(41)
model = Model()
```
